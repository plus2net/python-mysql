## 2. Importing Modules & Setting Up ParagraphStyle

We can import `Paragraph` and `ParagraphStyle` and declare style properties.

```python
from reportlab.pdfgen import canvas
from reportlab.lib.styles import ParagraphStyle
from reportlab.platypus import Paragraph
from reportlab.lib.pagesizes import A4

my_Style = ParagraphStyle(
    'My Para style',
    fontName='Times-Roman',
    backColor='#F1F1F1',
    fontSize=16,
    borderColor='#FFFF00',
    borderWidth=2,
    borderPadding=(20, 20, 20),
    leading=20,
    alignment=0
)
print("ParagraphStyle initialized!")
```

## 3. Drawing Styled Paragraph on Canvas

Connect style properties to our paragraph and draw on canvas with `wrapOn` and `drawOn`.

```python
width, height = A4
output_pdf = 'pdf_paragraph_styled.pdf'
c = canvas.Canvas(output_pdf, pagesize=A4)

p1 = Paragraph('''<b>About this online classes </b><BR/>\
     Welcome to our online classes.<BR/> \
	 You can learn Python, PHP, database like MySQL \
SQLite and many more \
<font face="times" color="blue">We are creating \
PDF by using ReportLab</font> \
<i>This is part of our Final report.</i>''', my_Style)

p1.wrapOn(c, 300, 50)
p1.drawOn(c, width - 450, height - 350)
c.save()

print(f"Styled PDF created: {output_pdf}")
```

## 4. Example 1: Handling Long Text with Automatic Page Breaks (SimpleDocTemplate)

We can easily manage content across multiple pages by using `SimpleDocTemplate` along with flowables like `Paragraph` and `Spacer`.

```python
from reportlab.lib.pagesizes import letter
from reportlab.lib.styles import getSampleStyleSheet
from reportlab.platypus import SimpleDocTemplate, Paragraph, Spacer

doc_pdf = 'pdf_long_text_doc.pdf'
doc = SimpleDocTemplate(doc_pdf, pagesize=letter)

styles = getSampleStyleSheet()
style = styles["Normal"]

long_text = """
plus2net.com is an invaluable resource for programmers and students looking to enhance their skills 
in various web programming languages. 

The site offers a comprehensive Python section that covers a wide range of topics, 
from basic syntax and data types to advanced concepts like decorators, generators, and context managers. 

Detailed tutorials, practical examples, and clear explanations make it easy for learners at all levels 
to understand and apply Python programming in real-world scenarios. 

Whether you are just starting out or looking to refine your skills, plus2net.com's Python section 
is designed to provide the knowledge and tools you need to succeed.
""" * 5

elements = []
for paragraph in long_text.split("\n"):
    if paragraph.strip():
        elements.append(Paragraph(paragraph, style))
        elements.append(Spacer(1, 12))

doc.build(elements)
print(f"Multi-page PDF created: {doc_pdf}")
```

## 5. Example 2: Handling Long Text with Canvas & Page Break Checks

Here by using canvas and Paragraph objects we will ensure that text is neatly divided across pages.

```python
from reportlab.lib.pagesizes import letter
from reportlab.pdfgen import canvas
from reportlab.lib.styles import getSampleStyleSheet
from reportlab.platypus import Paragraph

canvas_pdf = 'pdf_canvas_long.pdf'
c = canvas.Canvas(canvas_pdf, pagesize=letter)

styles = getSampleStyleSheet()
style = styles["Normal"]

long_text = """
plus2net.com is an invaluable resource for programmers and students looking to enhance their skills 
in various web programming languages. 

The site offers a comprehensive Python section covering basic syntax, data types, and advanced concepts.
""" * 5

page_width, page_height = letter
left_margin, right_margin, top_margin, bottom_margin = 50, 50, 50, 50
usable_width = page_width - left_margin - right_margin
usable_height = page_height - top_margin - bottom_margin

x = left_margin
y = page_height - top_margin

for paragraph_text in long_text.split("\n"):
    if paragraph_text.strip():
        paragraph = Paragraph(paragraph_text, style)
        width, height = paragraph.wrap(usable_width, usable_height)
        if y - height < bottom_margin:
            c.showPage()
            y = page_height - top_margin
        paragraph.drawOn(c, x, y - height)
        y -= height + 12

c.save()
print(f"Canvas multi-page PDF generated: {canvas_pdf}")
```
